```python
# ==========================================
# BANK ACCOUNT MANAGEMENT SYSTEM
# ==========================================

# ---------- Load Customer Information ----------

name = input("Enter customer name: ").strip().title()
age = int(input("Enter age: "))
income = float(input("Enter annual income: "))
credit_score = int(input("Enter credit score: "))
balance = float(input("Enter opening balance: "))

active = True


# ---------- Functions ----------

def show_menu():
    print("\n========== BANK MENU ==========")
    print("1. View Profile")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Loan Check")
    print("5. Security Test")
    print("6. Account Review")
    print("7. Exit")
    print("===============================")


def show_account_details(name, balance):
    print("\n---------- ACCOUNT PROFILE ----------")
    print(f"Customer Name : {name}")
    print(f"Balance       : ₹{balance:.2f}")
    print("-------------------------------------")


def deposit(balance, amount):
    if amount <= 0:
        print("Deposit amount must be greater than 0.")
    else:
        balance = balance + amount
        print(f"₹{amount:.2f} deposited successfully.")

    return balance


def withdraw(balance, amount):
    if amount <= 0:
        print("Withdrawal amount must be greater than 0.")
    elif amount > balance:
        print("Insufficient balance.")
    else:
        balance = balance - amount
        print(f"₹{amount:.2f} withdrawn successfully.")

    return balance


def check_loan_eligibility(age, income, credit, active):

    if age >= 21 and income >= 300000 and credit >= 700 and active:
        return True
    else:
        return False


def security_test():
    print("\n========== SECURITY TEST ==========")

    correct_password = "python123"

    for attempt in range(1, 4):

        print(f"Attempt {attempt} of 3")

        password = input("Enter password: ")

        if password == correct_password:
            print("Security test passed!")
            break

        print("Wrong password.")

    else:
        print("All attempts failed.")

    print("===================================")


def account_review():

    print("\n========== ACCOUNT REVIEW ==========")

    for day in range(1, 4):

        for check in range(1, 3):

            print(f"Day {day} - Review {check}: Account checked")

    print("====================================")


# ---------- Main Program ----------

print("\nWelcome to the Bank Management System")

while True:

    show_menu()

    choice = input("Enter your choice (1-7): ")

    if choice == "1":

        show_account_details(name, balance)

    elif choice == "2":

        amount = float(input("Enter deposit amount: "))
        balance = deposit(balance, amount)

    elif choice == "3":

        amount = float(input("Enter withdrawal amount: "))
        balance = withdraw(balance, amount)

    elif choice == "4":

        eligible = check_loan_eligibility(
            age,
            income,
            credit_score,
            active
        )

        if eligible:
            print("\nCongratulations!")
            print("You are eligible for a loan.")
        else:
            print("\nSorry.")
            print("You are not eligible for a loan.")

    elif choice == "5":

        security_test()

    elif choice == "6":

        account_review()

    elif choice == "7":

        print("\nThank you for using the Bank Management System.")
        print("Goodbye!")
        break

    else:

       print("Invalid choice. Please select 1-7.")
```
