```python
customer_name = input("Enter customer name: ").strip().title()
age = int(input("Enter age: "))
account_number = int(input("Enter account number: "))
ifsc_code = input("Enter IFSC code: ")
balance = float(input("Enter initial balance: "))

is_account_active = True
monthly_salary = 300000
is_guarantor_available = True
good_credit = True
print ("*********Welcome to the Axis Bank*********")
print(f"Account details of customer: {customer_name}\n A/C number is : {account_number} \n IFSC code is : {ifsc_code} \n Account_balance is {balance}")

while True:
    print("Please select the option from below menu")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Exit")
    option = input("1: ")
    if option == "1":
        deposit_amount = float(input("Enter the amount to deposit: "))
        balance += deposit_amount
        print(f"Amount deposited successfully! New balance is: {balance}")
    elif option == "2":
        withdraw_amount = float(input("Enter the amount to withdraw: "))
        if withdraw_amount <= balance:
            balance -= withdraw_amount
            print(f"Amount withdrawn successfully! New balance is: {balance}")
        else:
            print("Insufficient balance!")
    elif option == "3":
        print(f"Your current balance is: {balance}")
    elif option == "4":
        print("Thank you for banking with us!")
        break
    else:
        print("Invalid option! Please try again.")
```
